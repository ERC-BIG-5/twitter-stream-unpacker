```python
import json
import os
import shutil
from csv import DictReader
os.chdir("/home/rsoleyma/projects/twitter-stream-unpacker")
from src.consts import BASE_DATA_PATH
from src.consts import BASE_MEDIA_FOLDER
```

```python
media_dir = BASE_MEDIA_FOLDER / "2024102000"
img_ids = []
imgs: dict[str, list[str]] = {}
for img in media_dir.glob("*"):
    img_id = img.name.split("_")[0]
    img_ids.append(img_id)
    imgs.setdefault(img_id, []).append(img.name)
# jd = json.dumps(img_ids)
```

```python
all_items = {}
for file in (BASE_DATA_PATH / "temp/find").glob("*"):
    data = json.load(file.open(encoding="utf-8"))
    all_items.update(data["items"])
```

```python
import requests
from src.util import get_post_text, find_url

def follow_url(url):
    try:
        response = requests.get(url, allow_redirects=True, timeout=10)
        return response.url  # This is the final URL after all redirects

        # If you want full redirect history:
        # return response.history  # List of all redirects
        # return [h.url for h in response.history] + [response.url]  # All URLs in chain
    except requests.RequestException as e:
        return f"Error: {e}"

for id, post in all_items.items():
    loc, post_ = post
    urls = find_url(get_post_text(post_))
    for u in urls:
        print(follow_url(u))
    print(urls)
```

```python
from src import util


def build_tasks(filtered_out: list[str]) -> tuple[list[dict], list[str]]:
    """
    returns the list of tasks and image names to keep
    """
    tasks = []
    keep_imgs = []
    for id, post in all_items.items():
        loc, post_ = post
        task = {
            "post_text": util.get_post_text(post_)
        }
        has_img = False
        for idx, img_id in enumerate(sorted(imgs[id])):
            if img_id not in filtered_out:
                task[f"image_{idx}"] = img_id
                has_img = True
                keep_imgs.append(img_id)
        if has_img:
            tasks.append(task)
    return tasks, keep_imgs


filtered_out = [line["img"] + ".jpg" for line in list(DictReader((BASE_MEDIA_FOLDER / "fulvia_filter.csv").open())) if line["out"] == "1"]
tasks, keep_imgs = build_tasks(filtered_out)
print(len(tasks), len(keep_imgs))

# > 489 589
```
